```python
import pandas as pd
from selenium.common.exceptions import NoSuchElementException
from selenium.webdriver.common.by import By

def get_price_numbers(element):
    """Получает только числовое значение цены"""
    price_selectors = [
        './div[@class="product-card__middle-wrap"]/div/span/ins',
        './/span[@class="price"]//ins',
        './/span[@class="price"]',
        './/*[contains(@class, "price")]',
    ]

    for selector in price_selectors:
        try:
            price_element = element.find_element(By.XPATH, selector)

            price_text = price_element.get_attribute("textContent") or \
                        price_element.get_attribute("innerText") or \
                        price_element.text

            if price_text:
                import re
                numbers = re.findall(r'[\d\s]+', price_text)
                if numbers:
                    clean_price = numbers[0].strip()
                    return clean_price

        except NoSuchElementException:
            continue

    return "Цена не найдена"

# Создаем списки для данных
names = []
prices = []
article_ids = []
links = []

# Собираем данные с улучшенным парсингом цены
for i in range(len(div_elements)):
    try:
        name_element = div_elements[i].find_element(By.XPATH, './a')
        link_element = div_elements[i].find_element(By.XPATH, './a')
        id_element = article_elements[i].get_attribute('data-nm-id')

        # Пробуем оба способа получения цены
        price_clean = get_clean_price(div_elements[i])
        price_numbers = get_price_numbers(div_elements[i])

        # Используем тот, который дал лучший результат
        final_price = price_clean if price_clean != "Цена не найдена" else price_numbers

        # Добавляем данные в списки
        names.append(name_element.get_attribute('aria-label') or name_element.text)
        links.append(link_element.get_attribute('href'))
        prices.append(final_price)
        article_ids.append(id_element)

    except Exception as e:
        print(f"Ошибка при обработке товара {i+1}: {str(e)}")
        names.append(f"Ошибка: товар {i+1}")
        links.append("")
        prices.append("")
        article_ids.append("")

# Создаем DataFrame
df = pd.DataFrame({
    'название': names,
    'цена': prices,
    'артикул': article_ids,
    'ссылка': links
})

# Выводим результат
print("=" * 100)
print("ТАБЛИЦА ТОВАРОВ")
print("=" * 100)

# Настройка отображения
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', 40)

print(df.to_string(index=True))

# Показываем статистику
print(f"\n📊 Статистика:")
print(f"Всего товаров: {len(df)}")
print(f"Товаров с ценой: {len(df[df['цена'] != 'Цена не найдена'])}")
print(f"Товаров без цены: {len(df[df['цена'] == 'Цена не найдена'])}")

# Выводим примеры цен для проверки
print(f"\n🔍 Примеры цен (первые 5):")
for i, price in enumerate(prices[:5]):
    print(f"  {i+1}. '{price}'")
```
